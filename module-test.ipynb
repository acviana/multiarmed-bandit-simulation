```python
# Reference: https://matplotlib.org/stable/gallery/lines_bars_and_markers/scatter_hist.html

# Lay out the gridspec for the figure
fig = plt.figure(figsize=(6, 3))
gs = fig.add_gridspec(
    1,
    2,
    width_ratios=(4, 1),
    left=0.1,
    right=0.9,
    bottom=0.1,
    top=0.9,
    wspace=0.025,
    hspace=0.05,
)
fig.suptitle("Bandit rewards after 1k steps")

# Scatter Plot
ax_left = fig.add_subplot(gs[0, 0])
ax_left.plot(bandit.observations, ".", alpha=0.4)
ax_left.plot(bandit.means, label="Mean")
ax_left.set_xlabel("Steps")
ax_left.set_ylabel("Reward")
ax_left.grid(True)
ax_left.legend()

# Histogram Plot
ax_right = fig.add_subplot(gs[0, 1], sharey=ax_left)
n, bins, patches = ax_right.hist(
    bandit.observations, edgecolor="black", orientation="horizontal", alpha=0.4
)
ax_right.tick_params(axis="y", left=False, labelleft=False)
ax_right.grid(True)
```

## Examples of `TestBed` Class

```python
testbed_figure_epsilon = 0.00
testbed_bandit_count = 10

bandits = [
    main.Bandit(mean=random.gauss(mu=0, sigma=1), stdev=1)
    for _ in range(testbed_bandit_count)
]
test_bed = main.TestBed(bandits=bandits, default_mean=0.0)
test_bed.run_trials(steps=1000, epsilon=testbed_figure_epsilon)
```

```python
[bandit.true_mean for bandit in test_bed.bandits]
```

```python
# Lay out the gridspec for the figure
fig = plt.figure(figsize=(6, 3))
gs = fig.add_gridspec(
    1,
    2,
    width_ratios=(4, 1),
    left=0.1,
    right=0.9,
    bottom=0.1,
    top=0.9,
    wspace=0.025,
    hspace=0.05,
)

fig.suptitle(
    rf"Testbed Individual Bandit Rewards After 1k steps ($\epsilon$={testbed_figure_epsilon})"
)
ax_left = fig.add_subplot(gs[0, 0])
ax_right = fig.add_subplot(gs[0, 1], sharey=ax_left)

prop_cycle = plt.rcParams["axes.prop_cycle"]
colors = prop_cycle.by_key()["color"]

for index, bandit in enumerate(test_bed.bandits):
    # Scatter Plot
    ax_left.plot(bandit.observations, ".", color=colors[index], alpha=0.4)
    ax_left.plot(
        bandit.means,
        color=colors[index],
        linestyle="--",
        label=rf"$\mu=${bandit.true_mean:.2f}",
    )
    ax_left.set_xlabel("Steps")
    ax_left.set_ylabel("Reward")
    ax_left.grid(True)
    ax_left.legend()

    # Histogram Plot
    n, bins, patches = ax_right.hist(
        bandit.observations,
        edgecolor="black",
        color=colors[index],
        orientation="horizontal",
        alpha=0.4,
    )
    ax_right.tick_params(axis="y", left=False, labelleft=False)
    ax_right.grid(True)
```

## Example of `run_experiment` Function

```python
experiment_figure_experiments = 2000
experiment_figure_steps = 1000
experiment_figure_bandits = 10

experiment_000 = main.run_experiments(
    bandit_count=experiment_figure_bandits,
    steps=experiment_figure_steps,
    experiments=experiment_figure_experiments,
    epsilon=0.0,
)
experiment_001 = main.run_experiments(
    bandit_count=experiment_figure_bandits,
    steps=experiment_figure_steps,
    experiments=experiment_figure_experiments,
    epsilon=0.01,
)
experiment_010 = main.run_experiments(
    bandit_count=experiment_figure_bandits,
    steps=experiment_figure_steps,
    experiments=experiment_figure_experiments,
    epsilon=0.1,
)
```

```python
fig, ax = plt.subplots()
ax.plot(experiment_000, label=rf"$\epsilon$={0.00}")
ax.plot(experiment_001, label=rf"$\epsilon$={0.01}")
ax.plot(experiment_010, label=rf"$\epsilon$={0.10}")

ax.set_title(
    f"Average Reward for {experiment_figure_bandits} bandit simulation"
    f"after {experiment_figure_experiments} "
    f"experiments of {experiment_figure_steps} steps"
)
# ax.set_ylim(0.6, 0.8)
ax.set_xlabel("Step")
ax.set_ylabel("Average Reward")
ax.legend()
ax.grid(True)
```
